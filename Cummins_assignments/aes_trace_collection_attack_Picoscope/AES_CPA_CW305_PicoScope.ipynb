```python
inv_Shift_row_lut = [0,5,10,15,4,9,14,3,8,13,2,7,12,1,6,11]
```

```python
def sbox_lookup(byte):
    # print(x,y)
    x = byte >> 4
    #This step is taking the first character
    # print("\n",hex(x))
    y = byte & 15
    #This step is taking the second character
    # print("\n",hex(y))
    print("\n Printing aes sbox",hex(aes_sbox[x][y]))
    return aes_sbox[x][y]
```

```python
def reverse_lookup(byte):
    x = byte >> 4
    y = byte & 15
    return reverse_aes_sbox[x][y]
```

```python
def inv_shift_row(s):
    inv_s = ''
    inv_s += s[0:2] + s[26:28] + s[20:22] + s[14:16]
    inv_s += s[8:10] + s[2:4] + s[28:30] + s[22:24]
    inv_s += s[16:18] + s[10:12] + s[4:6] + s[30:32]
    inv_s += s[24:26] + s[18:20] + s[12:14] + s[6:8]
    return inv_s
```

```python
def p_correlation(key_byte, byte_index, traces, plaintexts, ciphertexts):

    power_model = []

    for i, ciphertext in enumerate(ciphertexts):
        ind = inv_Shift_row_lut[byte_index]
        addkey_outp = int(ciphertext[2*byte_index : 2*(byte_index+1)], 16) ^ key_byte
        sbox_outp = reverse_lookup(addkey_outp)
        power_model.append((sbox_outp^int(ciphertext[2*ind : 2*(ind+1)], 16)).bit_count())

    correlations = []
    for i in range(len(traces)):
        measurements = traces[i]
        x = [power_model, measurements]
        coefficient = np.corrcoef(np.array(power_model), np.array(measurements))
        correlations.append(abs(coefficient[0][1]))

    return max(correlations), correlations
```

```python
def get_correct_key_byte(byte_index, traces, plaintexts, ciphertexts):
    max_p = 0
    best_key_guess = 0
    p_list = []
    for key_guess in range(256):
        if key_guess%10==0:
            print(key_guess, end=' ')
        (p, temp) = p_correlation(key_guess, byte_index, traces, plaintexts, ciphertexts)
        p_list.append(temp)

        if p > max_p:
            max_p = p
            best_key_guess = key_guess

    return best_key_guess, max_p, p_list
```

```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

ROUND10_KEY = "13111d7fe3944a17f307a78b4d2b30c5"

# Use active AES window, not full 5000 samples first
# Full active AES window from your trace:
START = 950
END   = 2350

# For only final round later, try:
# START = 1900
# END   = 2350

NTRACES = 10000

# ------------------------------------------------------------
# Convert traces once
# Your `traces` is sample-major: traces[sample][trace_index]
# Convert to trace-major: T[trace_index, sample]
# ------------------------------------------------------------

T_full = np.asarray(traces, dtype=np.float32).T
T = T_full[:NTRACES, START:END]

plaintexts_use  = plaintexts[:NTRACES]
ciphertexts_use = ciphertexts[:NTRACES]

print("Trace matrix used:", T.shape)

# ------------------------------------------------------------
# Build AES SBox / InvSBox
# Assumes your notebook already has aes_sbox as 16x16 list
# ------------------------------------------------------------

SBOX = np.array([x for row in aes_sbox for x in row], dtype=np.uint8)

INV_SBOX = np.zeros(256, dtype=np.uint8)
for i, v in enumerate(SBOX):
    INV_SBOX[v] = i

HW = np.array([bin(x).count("1") for x in range(256)], dtype=np.float32)

# This is the same final-round byte relation we used earlier
INV_SHIFT_ROW_LUT = np.array(
    [0, 5, 10, 15,
     4, 9, 14, 3,
     8, 13, 2, 7,
     12, 1, 6, 11],
    dtype=np.int32
)

def hex_to_u8_matrix(hex_list):
    return np.array(
        [[int(h[2*i:2*i+2], 16) for i in range(16)] for h in hex_list],
        dtype=np.uint8
    )

C = hex_to_u8_matrix(ciphertexts_use)

# ------------------------------------------------------------
# Normalize traces once
# ------------------------------------------------------------

def normalize_cols(X):
    X = X.astype(np.float32)
    X = X - X.mean(axis=0, keepdims=True)
    s = X.std(axis=0, ddof=1, keepdims=True)
    s[s == 0] = 1.0
    return X / s

Tn = normalize_cols(T)

def fast_cpa_byte_round10(byte_i):
    guesses = np.arange(256, dtype=np.uint8)

    # Model:
    # before final round byte = InvSBox(C[byte_i] XOR K10_guess)
    # after final round byte  = C[InvShiftRows(byte_i)]
    ind = INV_SHIFT_ROW_LUT[byte_i]

    before = INV_SBOX[np.bitwise_xor(C[:, byte_i][:, None], guesses[None, :])]
    after  = C[:, ind][:, None]

    H = HW[np.bitwise_xor(before, after)]   # shape: NTRACES x 256

    Hn = normalize_cols(H)

    # Vectorized correlation:
    # Hn.T = 256 x NTRACES
    # Tn   = NTRACES x NSAMPLES
    corr = (Hn.T @ Tn) / (NTRACES - 1)

    scores = np.max(np.abs(corr), axis=1)
    best_guess = int(np.argmax(scores))
    best_score = float(scores[best_guess])
    best_sample = int(np.argmax(np.abs(corr[best_guess]))) + START

    return best_guess, best_score, best_sample, corr, scores

# ------------------------------------------------------------
# Recover all 16 round-10 key bytes
# ------------------------------------------------------------

retrieved_key = []
corr_mat = []
byte_info = []

for byte_i in tqdm(range(16), desc="Fast round-10 CPA"):
    best, score, sample, corr, scores = fast_cpa_byte_round10(byte_i)

    correct = int(ROUND10_KEY[2*byte_i:2*byte_i+2], 16)
    rank = int(np.where(np.argsort(scores)[::-1] == correct)[0][0]) + 1

    retrieved_key.append(best)
    corr_mat.append(corr)
    byte_info.append((byte_i, best, correct, rank, score, sample))

    print(
        f"byte {byte_i:02d}: "
        f"best={best:02x}, correct={correct:02x}, "
        f"rank={rank}, corr={score:.5f}, sample={sample}"
    )

final_key_hex = "".join(f"{b:02x}" for b in retrieved_key)

print("\nRecovered round-10 key:", final_key_hex)
print("Expected  round-10 key:", ROUND10_KEY)
```

```python
# =============================================================
# CPA correlation visualization (runs after the CPA recovery cell)
# =============================================================
from pathlib import Path
from matplotlib.collections import LineCollection

PLOT_DIR = Path("cpa_correlation_plots")
PLOT_DIR.mkdir(parents=True, exist_ok=True)

EXPECTED_BYTES = list(bytes.fromhex(ROUND10_KEY))

assert len(corr_mat) == len(retrieved_key) == len(EXPECTED_BYTES) == 16, \
    "Run the 'Fast round-10 CPA' cell first to fill all 16 byte results."
assert all(np.asarray(c).shape[0] == 256 for c in corr_mat), \
    "Each corr_mat[byte] should contain correlation traces for 256 guesses."


def _byte_corr(byte_i):
    """Return 256 x samples absolute Pearson correlations for byte_i."""
    return np.abs(np.asarray(corr_mat[byte_i], dtype=np.float32))


def _plot_byte_traces(ax, byte_i, show_legend=False):
    """Absolute correlation vs trace sample for all 256 hypotheses."""
    corr_abs = _byte_corr(byte_i)
    x = np.arange(START, START + corr_abs.shape[1])
    correct = EXPECTED_BYTES[byte_i]
    recovered = int(retrieved_key[byte_i])

    # Draw the other hypotheses in a single lightweight matplotlib collection.
    # Exclude both highlighted keys (they can coincide).
    remaining = np.array([g for g in range(256)
                          if g != correct and g != recovered], dtype=int)
    segments = np.stack([
        np.broadcast_to(x, (len(remaining), len(x))),
        corr_abs[remaining],
    ], axis=-1)
    ax.add_collection(LineCollection(segments,
                                    colors=[(0.45, 0.48, 0.53, 0.13)],
                                    linewidths=0.5,
                                    label="Other key guesses"))

    # Correct known byte: most important curve.
    ax.plot(x, corr_abs[correct], color="crimson", linewidth=2.1,
            zorder=5, label=f"Correct: 0x{correct:02X}")
    peak_at = int(np.argmax(corr_abs[correct]))
    ax.scatter([x[peak_at]], [corr_abs[correct, peak_at]],
               color="crimson", s=20, zorder=6)

    # Show separately if the top CPA guess is NOT the correct key byte.
    if recovered != correct:
        ax.plot(x, corr_abs[recovered], color="darkorange",
                linestyle="--", linewidth=1.7, zorder=4,
                label=f"Recovered: 0x{recovered:02X}")

    ax.set_xlim(int(x[0]), int(x[-1]))
    ax.set_ylim(0, max(0.01, float(corr_abs.max()) * 1.12))
    ax.set_xlabel("Sample index")
    ax.set_ylabel("|Pearson correlation|")
    ax.grid(alpha=0.22)
    if show_legend:
        ax.legend(fontsize=8, loc="upper left")


def _plot_byte_guess_scores(ax, byte_i, show_legend=False):
    """Highest |correlation| anywhere in time, for each of 256 guesses."""
    scores = _byte_corr(byte_i).max(axis=1)
    guesses = np.arange(256)
    correct = EXPECTED_BYTES[byte_i]
    recovered = int(retrieved_key[byte_i])

    ax.plot(guesses, scores, color="slategray", linewidth=0.9, alpha=0.7)
    ax.scatter(guesses, scores, color="gray", s=6, alpha=0.55)
    ax.scatter([correct], [scores[correct]], color="crimson",
               s=48, zorder=5,
               label=f"Correct 0x{correct:02X}: {scores[correct]:.4f}")
    ax.axvline(correct, color="crimson", linewidth=1.0, alpha=0.45)

    if recovered != correct:
        ax.scatter([recovered], [scores[recovered]], color="darkorange",
                   marker="D", s=43, zorder=5,
                   label=f"Recovered 0x{recovered:02X}: {scores[recovered]:.4f}")
        ax.axvline(recovered, color="darkorange", linestyle="--", alpha=0.4)

    ax.set_xlim(-3, 258)
    ax.set_ylim(0, max(0.01, float(scores.max()) * 1.15))
    ax.set_xticks([0, 64, 128, 192, 255])
    ax.set_xticklabels(["00", "40", "80", "C0", "FF"])
    ax.set_xlabel("Key guess (hex)")
    ax.set_ylabel("Max |Pearson correlation|")
    ax.grid(alpha=0.22)
    if show_legend:
        ax.legend(fontsize=8, loc="upper right")


def _byte_title(byte_i):
    expected = EXPECTED_BYTES[byte_i]
    recovered = int(retrieved_key[byte_i])
    peak_r = float(_byte_corr(byte_i)[expected].max())
    outcome = "MATCH" if expected == recovered else "MISMATCH"
    return (f"Byte {byte_i:02d} | Correct 0x{expected:02X} | "
            f"Recovered 0x{recovered:02X} | r={peak_r:.4f} | {outcome}")
```

```python
SBOX = [x for row in aes_sbox for x in row]

RCON = [
    0x00, 0x01, 0x02, 0x04, 0x08,
    0x10, 0x20, 0x40, 0x80, 0x1b, 0x36
]

def xor_word(a, b):
    return [x ^ y for x, y in zip(a, b)]

def g(word, round_no):
    # RotWord
    t = word[1:] + word[:1]

    # SubWord
    t = [SBOX[x] for x in t]

    # XOR Rcon
    t[0] ^= RCON[round_no]

    return t

def previous_round_key(curr_key, round_no):
    # curr_key = 16 bytes = [w0, w1, w2, w3], each word is 4 bytes
    c0 = curr_key[0:4]
    c1 = curr_key[4:8]
    c2 = curr_key[8:12]
    c3 = curr_key[12:16]

    p3 = xor_word(c3, c2)
    p2 = xor_word(c2, c1)
    p1 = xor_word(c1, c0)
    p0 = xor_word(c0, g(p3, round_no))

    return p0 + p1 + p2 + p3

def round10_to_master_key(round10_key_hex):
    key = list(bytes.fromhex(round10_key_hex))

    # Go backwards: K10 -> K9 -> ... -> K0
    for r in range(10, 0, -1):
        key = previous_round_key(key, r)

    return bytes(key).hex()


# Use this if retrieved_key contains all 16 recovered bytes
round10_key_hex = "".join(f"{b:02x}" for b in retrieved_key)

print("Recovered Round 10 key:", round10_key_hex)
print("Recovered Master key  :", round10_to_master_key(round10_key_hex))
```
